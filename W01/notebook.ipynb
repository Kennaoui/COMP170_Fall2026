## Meet the Terminal

A **program** is just a sequence of instructions a computer carries out, one at a time, top to bottom. "Running code" simply means: hand those instructions over and let the computer execute them, literally, exactly as written (see above).

Before we can write programs, we need a way to talk to the computer directly, without icons or clicking. That's the **terminal**.

Here's the chain of command:

`You → Terminal → Shell (Bash) → Operating System → Hardware`

Every command you type is a tiny program in its own right. When you type `ls`, Bash goes and finds the `ls` program, runs it, and shows you the result. You've technically been "programming" since your very first command, you just didn't know it yet.

<div class="keyterm">
<strong>Shell:</strong> the program that reads what you type and tells the operating system what to do with it (on Mac/Linux, usually <strong>Bash</strong>).<br>
<strong>Terminal:</strong> the window you type into.<br>
<strong>Prompt (<code>$</code>):</strong> the shell's way of saying "I'm listening."
</div>

Let's try it. Follow along on your own machine, command by command.

````python
whoami
````

`whoami` prints the username you're currently logged in as. It works exactly the same on Windows, macOS, and Linux, no alternative needed.

````python
pwd
````

`pwd`: **p**rint **w**orking **d**irectory. It answers exactly one question: "where, precisely, am I right now?" Think of your filesystem as a big filing cabinet: folders inside folders inside folders. `pwd` just tells you which drawer is currently open.

On Windows PowerShell, use `Get-Location` instead (some setups also accept `pwd` as a shortcut, but `Get-Location` is the one that always works).

````python
!ls
````

`ls` lists everything, files and folders alike, sitting in your current directory.

On Windows, the classic equivalent is `dir`. (PowerShell actually accepts `ls` too, as a shortcut for its own `Get-ChildItem`, but `dir` is the safer one to remember.)

````python
!ls -l
````

`ls -l` is the same listing, just with more detail added: file size, permissions, and when each item was last modified.

On Windows, plain `dir` already shows most of that detail by default, so there's no extra flag to remember.

````python
!mkdir -p comp170
%cd comp170
!pwd
````

`mkdir` creates a new folder, here one called `comp170`, and `cd` moves you into it. Both exist on Windows, macOS, and Linux alike; only the terminal you're typing them into changes.

On Windows, drop the `-p` flag: plain `mkdir comp170` does the same job. (The `-p` flag just tells Unix-style `mkdir` "don't complain if this folder already exists," which Windows' version doesn't need to be told.)

Congratulations: you just gave your computer a string of orders and it obeyed, no complaints, no snark. Don't get used to it: by Wednesday, Python is going to complain at you plenty. (That's not a bug. That's the whole point.)

## Vim: Your Tiny, Stubborn Text Editor

To write a program, we need a text editor: something that saves *plain text*, with zero hidden formatting. A Word document looks simple but is secretly stuffed with fonts, margins, and invisible styling, great for a résumé, useless for code a computer needs to read directly. Python doesn't want a résumé. It wants plain text.

Enter **Vim**: a text editor that has existed, mostly unchanged in spirit, since before most of your professors were born. It's available on almost every Unix-like system on Earth, it's fast once you know it, and it has a personality. Vim will not do anything you haven't explicitly told it to do. Type without switching modes first, and it just... judges you silently. Let's make sure that doesn't happen to you.

<table class="cmp-table">
<tr><th>Normal Mode</th><th>Insert Mode</th></tr>
<tr>
<td><strong>Enter with:</strong> <code>Esc</code></td>
<td><strong>Enter with:</strong> <code>i</code></td>
</tr>
<tr>
<td>Move around, delete, save, quit. Typing here does <em>not</em> insert text: every key is a command.</td>
<td>Type text normally, like in any other editor.</td>
</tr>
</table>

The golden rule of Vim survival: **when in doubt, press `Esc`.** It always brings you back to Normal mode, and it has never once broken anything.

We'll do the next part live, together, in an actual terminal. Here's what it looks like:

```
$ vim hello.py
i                        ← enter Insert mode
print("Hello, World!")
Esc                      ← back to Normal mode
:wq                      ← save and quit
```

<table class="cmp-table">
<tr><th>If this happens to you...</th><th>...try this</th></tr>
<tr><td>"I can't type anything!"</td><td>You're in Normal mode. Press <code>i</code>.</td></tr>
<tr><td>"I can't get out!"</td><td>Press <code>Esc</code>, then type <code>:q!</code> to quit without saving.</td></tr>
<tr><td>"I saved it but I don't know where it went."</td><td>Run <code>ls</code> to check what's there, <code>pwd</code> to confirm where you are.</td></tr>

## Naming Things: Variables

A **variable** is a labeled box for a value: Python's version of writing a name on a sticky note and slapping it on a container, so future-you (or the next line of code) knows what's inside without re-reading the whole recipe.

````python
name = "Leo"
print("Hello,", name)
print("Welcome to COMP 170.")
````

Walking through it, one line at a time:

1. Line 1 stores the text `"Leo"` under the name `name`; nothing is printed yet.
2. Line 2 calls `print` with two things; Python quietly adds a space between them.
3. Line 3 prints a second, independent line.

Python runs top to bottom, one line at a time, no skipping ahead and no going back, the same literal-minded discipline as your non-improvising, recipe-following cook from earlier this week.

## The Edit-Run Loop

Here's a small secret about what programming actually feels like, day to day: you change one thing, you run it, you look at what happened, and you adjust. Over and over.

`edit → save → run → read the output → repeat`

It's the same instinct as tasting a sauce, deciding it needs more salt, adding a pinch, tasting again. Nobody gets a program right on the first try, professionals included. The loop *is* the job, in every language, forever.
